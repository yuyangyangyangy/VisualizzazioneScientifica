```python
import pandas as pd
import matplotlib.pyplot as plt

# Carica e prepara i dati
df = pd.read_csv('Genere film - Foglio1.csv')
df.set_index('Nome film', inplace=True)
df_filled = df.fillna(0).astype(int)

# Calcola le somme per genere
genre_counts = df_filled.sum()

# Crea la figura
plt.figure(figsize=(14, 8))

# Crea l'istogramma con matplotlib
bars = plt.bar(range(len(genre_counts)), genre_counts.values, 
               color=plt.cm.viridis(range(len(genre_counts))), 
               edgecolor='black', linewidth=1)

# Aggiungi i valori sopra ogni barra
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 0.1,
             f'{int(height)}', ha='center', va='bottom', fontsize=10, fontweight='bold')

# Configura titolo e assi
plt.title('Distribuzione dei Generi Cinematografici', fontsize=16, pad=20)
plt.xlabel('Generi', fontsize=12)
plt.ylabel('Numero di Film', fontsize=12)
plt.xticks(range(len(genre_counts)), genre_counts.index, rotation=45, ha='right')

# Aggiungi griglia per migliore leggibilità
plt.grid(axis='y', alpha=0.3, linestyle='--')

# Imposta limiti dell'asse Y
plt.ylim(0, max(genre_counts.values) + 1)

# Aggiusta il layout
plt.tight_layout()

# Mostra l'istogramma
plt.show()
```
