```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Carica il dataset
df = pd.read_csv('Genere film - Foglio1.csv')

# Imposta i nomi dei film come indice
df.set_index('Nome film', inplace=True)

# Sostituisce i valori NaN con 0 e converti in interi
df_filled = df.fillna(0).astype(int)

# Crea la figura con dimensioni adeguate
plt.figure(figsize=(12, 8))

# Crea la heatmap
sns.heatmap(df_filled, 
            annot=True, 
            cmap='YlOrRd', 
            cbar_kws={'label': 'Presenza Genere'},
            linewidths=0.5,
            linecolor='gray')

# Configura titolo e assi
plt.title('Presenza dei Generi nei Film', fontsize=16, pad=20)
plt.xlabel('Generi', fontsize=12)
plt.ylabel('Film', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)

# Aggiusta il layout
plt.tight_layout()

# Mostra la heatmap
plt.show()
```
